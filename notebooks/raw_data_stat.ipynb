```python
import csv
import gzip

import matplotlib.pyplot as plt

data_path = "../data/causal_react_raw.csv.gz"
```

```python
# Viz seconds distribution in hist
seconds = []
dataset_set = set()
with gzip.open(data_path, "rt", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    # show cols
    print(f"columns: {reader.fieldnames}")
    for row in reader:
        seconds.append(float(row["seconds"]))
        dataset_set.add(row["src_set"])

print(f"unique sets: {dataset_set}")
print(f"count: {len(seconds)}")
print(f"min:   {min(seconds):.2f}")
print(f"max:   {max(seconds):.2f}")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(seconds, bins=100)
plt.yscale("log")
plt.xlabel("Seconds")
plt.ylabel("Count")
plt.title("Video Duration Distribution")
plt.tight_layout()
plt.show()
```
